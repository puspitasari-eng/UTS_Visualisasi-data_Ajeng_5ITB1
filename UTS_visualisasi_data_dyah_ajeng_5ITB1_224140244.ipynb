**1. Grafik batang**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files
uploaded = files.upload()

# Membaca data dari Sheet 'Soal Nomor 1'
df_soal1 = pd.read_excel('Data Kasus Visualisasi.xlsx', sheet_name='Soal Nomor 1', skiprows=3, nrows=4)
df_soal1 = df_soal1.iloc[:, :8]
df_soal1.columns = ['No', 'Kategori_Sepatu', 'Januari', 'Februari', 'Maret', 'April', 'Mei', 'Juni']

# Menghitung total penjualan per merek sepatu selama 6 bulan
df_soal1['Total_Penjualan'] = df_soal1[['Januari', 'Februari', 'Maret', 'April', 'Mei', 'Juni']].sum(axis=1)

# Visualisasi Grafik Batang dengan Warna Berbeda tiap Kategori
plt.figure(figsize=(9, 5))
sns.barplot(data=df_soal1, x='Kategori_Sepatu', y='Total_Penjualan', palette='Set2')
plt.title('Total Penjualan Kategori Sepatu (Januari - Juni)', fontsize=14, fontweight='bold')
plt.xlabel('Kategori Sepatu', fontsize=12)
plt.ylabel('Total Penjualan (Unit)', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

**pesan yang terkandung dalam grafik batang**
*   **Dominasi Pasar Merek:** Berdasarkan total akumulasi penjualan dari bulan Januari hingga Juni, merek Adidas mencatatkan volume penjualan tertinggi dibandingkan merek lainnya, menunjukkan tingkat minat dan permintaan konsumen yang paling besar terhadap merek tersebut.
*  **Evaluasi Performa Penjualan:** Melalui visualisasi warna yang berbeda-beda pada setiap batang kategori merek (Nike, Puma, Adidas, Converse), kita dapat dengan mudah mengidentifikasi merek yang memiliki penjualan paling rendah sehingga memerlukan strategi promosi, diskon, atau evaluasi stok tambahan pada periode berikutnya.

**2. Grafik garis**

```python
from google.colab import files
uploaded = files.upload()

# Membaca data dari Sheet 'Soal Nomor 2'
df_soal2 = pd.read_excel('Data Kasus Visualisasi.xlsx', sheet_name='Soal Nomor 2', skiprows=2, nrows=12)
df_soal2 = df_soal2.iloc[:, :4]
df_soal2.columns = ['Bulan', 'Produk_A', 'Produk_B', 'Produk_C']

# Visualisasi Grafik Garis dengan Warna Berbeda pada Setiap Produk
plt.figure(figsize=(11, 6))
plt.plot(df_soal2['Bulan'], df_soal2['Produk_A'], marker='o', color='#1f77b4', linewidth=2.5, label='Produk A')
plt.plot(df_soal2['Bulan'], df_soal2['Produk_B'], marker='s', color='#ff7f0e', linewidth=2.5, label='Produk B')
plt.plot(df_soal2['Bulan'], df_soal2['Produk_C'], marker='^', color='#2ca02c', linewidth=2.5, label='Produk C')

plt.title('Perkembangan Penjualan Produk A, B, dan C Selama 1 Tahun', fontsize=14, fontweight='bold')
plt.xlabel('Bulan', fontsize=12)
plt.ylabel('Jumlah Penjualan (Unit)', fontsize=12)
plt.xticks(rotation=45)
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.tight_layout()
plt.show()
```

**pesan yang terkandung dalam grafik garis**
*  **Tren Pertumbuhan Positif:** Grafik garis menunjukkan pergerakan tren penjualan yang konsisten mengalami kenaikan dari bulan Januari hingga Desember untuk ketiga produk, yang mengindikasikan ekspansi pasar yang sehat dan stabil sepanjang tahun.
*   **Konsistensi Keunggulan Produk:** Produk A secara konsisten berada di posisi teratas dalam hal jumlah unit terjual setiap bulannya dibandingkan Produk B dan Produk C, menjadikannya produk utama (core product) yang paling diandalkan dalam menyumbang angka penjualan

**3. Grafik lingkaran**

```python
from google.colab import files
uploaded = files.upload()

# Membaca data dari Sheet 'Soal Nomor 3'
df_soal3 = pd.read_excel('Data Kasus Visualisasi.xlsx', sheet_name='Soal Nomor 3', skiprows=2)
# Mengambil kolom data responden yang relevan
df_responden = df_soal3[['Jenis Kelamin', 'Usia', 'Status']].dropna().astype(int)

# Mapping Kategori (berdasarkan keterangan pada file Excel)
df_responden['Jenis_Kelamin_Label'] = df_responden['Jenis Kelamin'].map({1: 'Laki-laki', 2: 'Perempuan'})
df_responden['Usia_Label'] = df_responden['Usia'].map({
    1: '< 21 Th', 2: '21-30 Th', 3: '31-40 Th', 4: '41-50 Th', 5: '> 50 Th'
})
df_responden['Status_Label'] = df_responden['Status'].map({1: 'Belum Kawin', 2: 'Kawin'})

# Visualisasi 3 Grafik Lingkaran (Pie Chart)
fig, axes = plt.subplots(1, 3, figsize=(16, 6))

# 1. Jenis Kelamin
jk_counts = df_responden['Jenis_Kelamin_Label'].value_counts()
axes[0].pie(jk_counts, labels=jk_counts.index, autopct='%1.1f%%', startangle=140, colors=['#3498db', '#e74c3c'])
axes[0].set_title('Distribusi Jenis Kelamin', fontsize=12, fontweight='bold')

# 2. Usia
usia_counts = df_responden['Usia_Label'].value_counts()
axes[1].pie(usia_counts, labels=usia_counts.index, autopct='%1.1f%%', startangle=140, colors=sns.color_palette('pastel'))
axes[1].set_title('Distribusi Kelompok Usia', fontsize=12, fontweight='bold')

# 3. Status
status_counts = df_responden['Status_Label'].value_counts()
axes[2].pie(status_counts, labels=status_counts.index, autopct='%1.1f%%', startangle=140, colors=['#2ecc71', '#f1c40f'])
axes[2].set_title('Distribusi Status Perkawinan', fontsize=12, fontweight='bold')

plt.tight_layout()
plt.show()
```

**pesan yang terkandung dalam grafik lingkaran**


*   **Distribusi Jenis Kelamin:** Grafik lingkaran pertama menunjukkan proporsi demografi responden/pegawai, di mana komposisi didominasi oleh salah satu gender (Laki-laki), yang memberikan gambaran jelas mengenai profil tenaga kerja atau target audiens di lingkungan instansi tersebut.
*   **Kelompok Usia Produktif:** Grafik lingkaran kedua memperlihatkan bahwa sebagian besar responden berada pada rentang usia produktif (seperti kelompok usia 21–30 tahun atau 31–40 tahun), mencerminkan tingkat usia yang aktif dan matang secara profesional.
*   **Status Perkawinan:** Grafik lingkaran ketiga menggambarkan proporsi status perkawinan responden (di mana mayoritas berstatus Kawin), yang dapat menjadi acuan penting dalam pertimbangan kebijakan internal, tunjangan, atau analisis sosial dari data responden tersebut.
